# Data CLeaning

```python
df_clean = df.drop_duplicates().copy()
```

```python
df_clean[df.isna().any(axis=1)]
```

```python
def bersihkan_nilai(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).lower().replace('poin', '').strip().replace(',', '.')
    try:
        angka = float(val_str)
        # Nilai 999 adalah kesalahan input data (skala nilai 0-100)
        if angka > 100:
            return np.nan
        return angka
    except ValueError:
        return np.nan
```

```python
df_clean['nilai'] = df_clean['nilai'].apply(bersihkan_nilai)
```

```python
df_clean['nilai'] = df_clean.groupby('mata_pelajaran')['nilai'].transform(lambda x: x.fillna(x.median()))
```

```python
df_clean['guru_pengampu'] = df_clean['guru_pengampu'].fillna('Belum Terdata')
```

```python
df_clean['jenis_ujian'] = df_clean['jenis_ujian'].str.upper()
```

```python
df_clean.duplicated().sum()
```

```python
# a. Kolom Turunan: Menentukan status kelulusan berdasarkan batas KKM (75)
KKM = 75
df_clean['status_kelulusan'] = np.where(df_clean['nilai'] >= KKM, 'Tuntas', 'Remedial')
```

# Data Manipulation

```python
# b. Sorting: Mengurutkan dataset berdasarkan nilai tertinggi ke terendah
df_sorted = df_clean.sort_values(by='nilai', ascending=False)
print(df_sorted[['nama', 'kelas', 'mata_pelajaran', 'nilai', 'status_kelulusan']].head())
```

```python
df_sorted
```

```python
# c. Filtering: Menyaring data siswa yang harus mengikuti remedial
remedial_list = df_clean[df_clean['status_kelulusan'] == 'Remedial']
print(f"\nJumlah ujian yang berstatus remedial: {len(remedial_list)} siswa.")
```

```python
total_data = len(df_clean)
jumlah_remedial = len(remedial_list)
jumlah_tuntas = total_data - jumlah_remedial

persen_tuntas = (jumlah_tuntas / total_data) * 100
persen_remedial = (jumlah_remedial / total_data) * 100

print(f"Persentase Tuntas   : {persen_tuntas:.2f}% ({jumlah_tuntas} siswa)")
print(f"Persentase Remedial : {persen_remedial:.2f}% ({jumlah_remedial} siswa)")
```
